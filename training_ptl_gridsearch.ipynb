### Training with Grid Search

```python
# Define the grid search parameters
# Example grid search: varying batch size and learning rate
grid_params = {
    'batch_size': [16, 32, 64],
    'epochs': [300], # Keep epochs relatively low for grid search iterations
    'img_size': [640],
    'cfg': [config_file],
    'data': [data_yaml_path],
    'weights': ['./yolov7-tiny.pt'], # Start training from scratch or a pretrained tiny model
    'hyp': [hyperparameter_file1, hyperparameter_file2], # Use the default hyperparameter file for tiny
    'name': ['yolov7-tiny-ptl-gridsearch'],
    'opt-alg': ['sgd', 'adam'],
    'early-stop-patience': [10] # Early stopping patience
}
```

```python
# Function to train a single model configuration
def train_model(params):
    print("\n" + "="*50)
    print("Starting training with parameters:")
    print(json.dumps(params, indent=2))
    print("="*50 + "\n")

    # Construct the training command
    command = [
        'python', 'train.py',
        '--img-size', str(params['img_size']),
        '--batch-size', str(params['batch_size']),
        '--epochs', str(params['epochs']),
        '--data', params['data'],
        '--cfg', params['cfg'],
        '--weights', params['weights'],
        '--hyp', params['hyp'],
        '--name', params['name'],
        '--cache-images',
        '--early-stop-patience', str(params['early-stop-patience'])
    ]

    # Add --device if you want to specify GPU or CPU
    command.extend(['--device', '0']) # Use GPU 0
    # command.extend(['--device', 'cpu']) # Use CPU

    # Add --bbox_interval and --image_weights if using wandb and logging specific things
    # command.extend(['--bbox_interval', '1', '--image_weights'])

    # Optimization Alg
    if params['opt-alg'] == 'adam':
        command.extend(['--adam'])

    print(f"Running: {command}")
    process = subprocess.run(command)

    if process.returncode != 0:
        print(f"Training failed with return code {process.returncode}")
    else:
        print("Training completed successfully.")

    # try:
    #     # Execute the training command
    #     process = subprocess.Popen(command, stdout=subprocess.PIPE, stderr=subprocess.PIPE)
    #     stdout, stderr = process.communicate()

    #     print("STDOUT:")
    #     print(stdout.decode('utf-8'))
    #     print("STDERR:")
    #     print(stderr.decode('utf-8'))

    #     if process.returncode != 0:
    #         print(f"Training failed with return code {process.returncode}")
    #     else:
    #         print("Training completed successfully.")

    # except Exception as e:
    #     print(f"An error occurred during training: {e}")
```

```python
# Simple grid search implementation
# This is a basic implementation. For more complex grid searches,
# consider libraries like sklearn.model_selection.ParameterGrid
# or tools integrated with experiment trackers like Wandb Sweeps.
from itertools import product

# Generate combinations of parameters
param_combinations = list(product(
    grid_params['batch_size'],
    grid_params['epochs'],
    grid_params['img_size'],
    grid_params['cfg'],
    grid_params['data'],
    grid_params['weights'],
    grid_params['hyp'],
    grid_params['name'],
    grid_params['opt-alg'],
    grid_params['early-stop-patience']
))

print(f"Total grid search runs: {len(param_combinations)}")
```

```python
# combo position to continue from
continue_from_combo = 0

# Run 1 training for each combination
for i, combo in enumerate(param_combinations):

    # skip combo
    if i < continue_from_combo:
        continue

    model_name = f"{combo[7]}_{i}" # Append index to name for unique runs
    print(f"Trainning start: {model_name}")
    params = {
        'batch_size': combo[0],
        'epochs': combo[1],
        'img_size': combo[2],
        'cfg': combo[3],
        'data': combo[4],
        'weights': combo[5],
        'hyp': combo[6],
        'name': model_name, 
        'opt-alg': combo[8],
        'early-stop-patience': combo[9]
    }
    print(params)
    train_model(params)
    print(f"Trainning finished: {model_name}")

print("\n" + "="*50)
print("Grid search completed.")
print("="*50)
```
